## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from inactivity_engine import InactivityDisambiguationEngine
from data_pipeline import get_train_val_test_splits, generate_student_history
from model import MentalHealthTwinPredictor

print("[OK] All libraries successfully imported.")
```

## 2. Load & Prepare StudentLife Data Splits

```python
splits = get_train_val_test_splits(n_samples=300, seed=42)
train_data = splits["train"]
val_data = splits["val"]
test_data = splits["test"]

print(f"Train Static Shape   : {train_data['static'].shape} (Demographics + Baseline Traits)")
print(f"Train Temporal Shape : {train_data['temporal'].shape} (30 days x 10 sensing streams)")
print(f"Train Targets Shape  : {train_data['targets'].shape} (Stress, Anxiety, Depression, Burnout)")
print(f"Test Set Count       : {len(test_data['targets'])} students")
```

## 3. The Core Novelty: Inactivity Disambiguation Engine
Testing two opposite situations where phone screen time is zero:

```python
engine = InactivityDisambiguationEngine()

# Case 1: Healthy Digital Detox (Outdoor, walking, daylight)
res_detox = engine.evaluate_inactivity_context(
    screen_time_hours=0.8, phone_unlocks=12, daytime_dark_hours=0.2,
    step_count=6500, ambient_conversation_hours=2.5, campus_mobility_entropy=0.75
)
print("Case 1 (Healthy Digital Detox):")
print(f"  State: {res_detox['state']}")
print(f"  Withdrawal Index: {res_detox['withdrawal_index']}")
print(f"  Rationale: {res_detox['clinical_rationale']}\n")

# Case 2: Depressive Bed-Bound Withdrawal (Dark room, 0 steps)
res_withdraw = engine.evaluate_inactivity_context(
    screen_time_hours=0.4, phone_unlocks=5, daytime_dark_hours=6.5,
    step_count=80, ambient_conversation_hours=0.0, campus_mobility_entropy=0.05
)
print("Case 2 (Depressive Bed-Bound Withdrawal):")
print(f"  State: {res_withdraw['state']}")
print(f"  Withdrawal Index: {res_withdraw['withdrawal_index']}")
print(f"  Rationale: {res_withdraw['clinical_rationale']}")
```
